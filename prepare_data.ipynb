Load the data

```python
import numpy as np
from sklearn.datasets import load_digits
```

```python
digits=load_digits()

x=digits.images
y=digits.target
```

```python
print('Images shape:', x.shape)
print('Labels shape:', y.shape)
```

```python
print("Minimum pixel value:", x.min())
print("Maximum pixel value:", x.max())

print("Available digits:", np.unique(y))
```

```python

# Normalize pixel values between 0 and 1
x = x.astype(np.float32) / 16.0

# Check normalized values
print("After normalization:")
print("Minimum pixel:", x.min())
print("Maximum pixel:", x.max())
print("Data type:", x.dtype)
```

Split data into training, validation, and test sets

```python
from sklearn.model_selection import train_test_split


# First split: 70% training and 30% temporary data

x_train, x_temp, y_train, y_temp = train_test_split(
    x,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

# Second split: 15% validation and 15% testing

x_val, x_test, y_val, y_test = train_test_split(
    x_temp,
    y_temp,
    test_size=0.50,
    random_state=42,
    stratify=y_temp
)

# Display dataset sizes

print("\nDataset splitting results:")

print("Training images:", x_train.shape[0])
print("Validation images:", x_val.shape[0])
print("Testing images:", x_test.shape[0])
```

Save prepared datasets to .npz files

```python

# Save prepared datasets
np.savez_compressed(
    "data/digit_splits.npz",

    X_train=x_train,
    y_train=y_train,

    X_val=x_val,
    y_val=y_val,

    X_test=x_test,
    y_test=y_test
)

print("\nAll datasets saved successfully!")
```
